```python
import pandas as pd
import joblib
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering
```

```python
df = pd.read_csv("smartcartclean.csv")
```

```python
cat_cols = ["Education","Living_With"]
ohe = OneHotEncoder(
    handle_unknown = "ignore"
)
enc_cols = ohe.fit_transform(df[cat_cols])
enc_df = pd.DataFrame(enc_cols.toarray(),columns = ohe.get_feature_names_out(cat_cols),index = df.index)
```

```python
X = pd.concat([df.drop(columns = cat_cols),enc_df],axis = 1)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
pca = PCA(n_components = 3)
X_pca = pca.fit_transform(X_scaled)
```

```python
agg_clf = AgglomerativeClustering(
    n_clusters = 4,
    linkage = "ward"
)
labels = agg_clf.fit_predict(X_pca)
```

```python
joblib.dump(ohe,"ohe.pkl")
joblib.dump(scaler,"scaler.pkl")
joblib.dump(pca,"pca.pkl")
joblib.dump(agg_clf,"agg_clf.pkl")
```

```python
df.to_csv("final.csv")
```

```python
df_new = pd.read_csv("final.csv")
```

```python
clusters_labels = df_new["clusters"].values
```

```python
import numpy as np
cluster_centers = {}

for cluster in np.unique(clusters_labels):
    cluster_centers[cluster] = X_pca[clusters_labels == cluster].mean(axis = 0) # axis = 0 har column ka alag alag mean nikalo 1 may har row ka alag alag mean
    
```

```python
joblib.dump(cluster_centers,"centers.pkl")
```
